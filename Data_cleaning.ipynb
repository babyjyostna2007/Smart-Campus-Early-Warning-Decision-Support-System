<h2 style="color: darkgreen;">Required Installations</h2>

```python
import pandas as pd
import numpy as np
```

<h2 style="color: darkblue;">📊 Load Dataset</h2>

```python
df = pd.read_csv("smart_campus.csv")
```

```python
print("Original Shape:", df.shape)
```

<h2 style="color: darkyellow;">📊 Check Missing Values</h2>

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

<h2 style="color: darkpurple;">📊 Remove Duplicates</h2>

```python
df = df.drop_duplicates()
```

<h2 style="color: darkred;">📊 Handle Missing Numerical Values</h2>

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

<h2 style="color: darkorange;">📊 Convert Categorical Missing Values</h2>

```python
df["Risk_Level"] = df["Risk_Level"].fillna("Medium")
df["Risk_Factors"] = df["Risk_Factors"].fillna("No major risk factors")
df["Recommended_Action"] = df["Recommended_Action"].fillna(
    "Continue regular monitoring"
)
```

<h2 style="color: darkred;">📊 Check Invalid Percentage Values</h2>

```python
percentage_columns = [
    "Attendance",
    "Assignment_Rate",
    "Occupancy_Rate",
    "Internet_Usage",
    "Humidity",
    "Water_Level"
]

for col in percentage_columns:
    df[col] = df[col].clip(0, 100)
```

<h2 style="color: darkblue;">📊 Values that cannot be negative</h2>

```python
non_negative_columns = [
    "Electricity_Usage",
    "Rainfall",
    "Maintenance_Complaints",
    "Backlogs",
    "Air_Quality_Index"
]

for col in non_negative_columns:
    df[col] = df[col].clip(lower=0)
```

<h2 style="color: darkpurple;">📊 Outlier Analysis</h2>

```python
outlier_columns = [
    "Attendance",
    "Current_GPA",
    "Previous_GPA",
    "Assignment_Rate",
    "Backlogs",
    "Occupancy_Rate",
    "Electricity_Usage",
    "Internet_Usage",
    "Maintenance_Complaints",
    "Temperature",
    "Humidity",
    "Rainfall",
    "Water_Level",
    "Air_Quality_Index"
]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(
        f"{col}: {len(outliers)} outliers"
    )
```

<h2 style="color: darkred;">📊 Save Cleaned Dataset</h2>

```python
df.to_csv("cleaned_smart_campus.csv", index=False)

print("\nCleaning completed!")
print("Final Shape:", df.shape)
print("\nRemaining Missing Values:")
print(df.isnull().sum())
```
